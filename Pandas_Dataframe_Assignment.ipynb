```python
#Q2. Data Inspection
    #Perform the following inspection tasks:
    #1.Display the first 5 rows of the dataset.
    #2.Display the last 5 rows.
    #3.Show the structure of the dataset using info().
    #4.Display the shape of the dataset.
#ANSWER:-
# 1. Display first 5 rows
print("First 5 Rows:")
print(df.head())

# 2. Display last 5 rows
print("\nLast 5 Rows:")
print(df.tail())

# 3. Show dataset structure using info()
print("\nDataset Info:")
df.info()

# 4. Display dataset shape
print("\nDataset Shape:")
print(df.shape)
```

```python
#Q3. Column Selection
     #Select and display only the following columns:
     #Product
     #Region
     #Sales
#ANSWER:-
# Load dataset
df = pd.read_csv("Retail_dataset.csv")
pd.set_option('display.max_rows', None)

# Select only Product, Region, Sales columns
selected_columns = df[["Product", "Region", "Sales"]]

# Display result
print(selected_columns)
```

```python
#Q4. Indexing using .loc
    #Using .loc, perform the following:
    #1. Display the row with index 10.
    #2.Display the Sales value of the row with index 20

#ANSWER:-

# 1. Display row with index 10
print("Row with index 10:")
print(df.loc[10])

# 2. Display Sales value of row with index 20
print("\nSales value at index 20:")
print(df.loc[20, "Sales"])
```

```python
#Q5. Indexing using .iloc
     #Using .iloc, display:
     #1.The first 6 rows and first 4 columns.
     #2.The Sales value from the 3rd row.
#ANSWER:-

# 1. Display first 6 rows and first 4 columns
print("First 6 rows and first 4 columns:")
print(df.iloc[0:6, 0:4])

# 2. Display Sales value from the 3rd row
print("\nSales value from the 3rd row:")
print(df.iloc[2]["Sales"])
```

```python
#Q6. Row Slicing
     #Display:
     #1.The first 10 rows of the dataset.
     #2.Rows 15 to 25.

#ANSWER:-

# 1. Display first 10 rows
print("First 10 Rows:")
print(df.head(10))

# 2. Display rows 15 to 25
print("\nRows 15 to 25:")
print(df.iloc[15:26])
```

```python
#Q7. Filtering Data
    #Filter the dataset to display:
    #1.Rows where Region = "East".
    #2.Rows where Sales > 1500.

#ANSWER:-

# 1. Rows where Region = "East"
print("Rows where Region = 'East':")
print(df[df["Region"] == "East"])

# 2. Rows where Sales > 1500
print("\nRows where Sales > 1500:")
print(df[df["Sales"] > 1500])
```

```python
#Q8. Multiple Condition Filtering
     #Display rows where:
     #1.Region is West
     #2.Sales is greater than 1200

#ANSWER:-

# Filter rows with multiple conditions
print("Rows where Region = 'West' and Sales > 1200:")

print(df[(df["Region"] == "West") & (df["Sales"] > 1200)])

```

```python
#Q9. Descriptive Statistics
     #Generate summary statistics using:
     #describe()
     #From the result identify:
     #.Mean sales value
     #.Maximum quantity sold
     #.Minimum sales value

#ANSWER:-
# Generate summary statistics
stats = df.describe()

print("Summary Statistics:")
print(stats)

# Mean sales value
print("\nMean Sales Value:")
print(stats.loc["mean", "Sales"])

# Maximum quantity sold
print("\nMaximum Quantity Sold:")
print(stats.loc["max", "Quantity"])

# Minimum sales value
print("\nMinimum Sales Value:")
print(stats.loc["min", "Sales"])
```

```python
#Q10. Aggregation Analysis
       #Using the Sales column, calculate:
       #1.Total sales
       #2.Average sales
       #3.Maximum sales
       #4.Minimum sales

#ANSWER:-
# Total sales
print("Total Sales:")
print(df["Sales"].sum())

# Average sales
print("\nAverage Sales:")
print(df["Sales"].mean())

# Maximum sales
print("\nMaximum Sales:")
print(df["Sales"].max())

# Minimum sales
print("\nMinimum Sales:")
print(df["Sales"].min())
```
